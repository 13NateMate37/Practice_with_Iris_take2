# Section 1 - Extraction and initial inspection 

```python
# Open the csv as a pandas dataframe

df_raw = pd.read_csv("Dataset/Raw/Iris.csv")

df_raw.head(10)
```

```python
# Insepcting the initial construction of the data

print("Initial shape:", df_raw.shape)
print('=' * 50)
print("Data types:\n", df_raw.dtypes)
print('=' * 50)
print("missing value count:\n", df_raw.isnull().sum())
print('=' * 50)
print("Duplicates count:\n", df_raw.duplicated().sum())
df_raw.describe()
      
```

The dataset is pretty clean with no missing values or duplicates. 

This should be really easy to work with.

# Section 2 - Transformation 

None of the columns datatypes are out of place for the information it's suppposed to be. 

The titles dont't particularly need work. I'm fine with camel case thought there seems to be a dispute on which you should use based on your language.

No missing values and no duplicates with an evenly distributed sample size.

```python
# Showing the distribution of the species in the dataset 
# Here are a few ways to do so

print("Total sample size:", df_raw['Species'].value_counts().sum())

print(df_raw.groupby("Species").size())


ax = df_raw["Species"].value_counts().plot(
    kind="bar",
    title="Species Distribution",
    xlabel="Species",
    ylabel="Count",
    color=["#FF9999", "#66B2FF", "#99FF99"]
)


plt.show()

# 150 / 3 = 50 samples per species.
```

We'll check our inital outliers

```python
# Outlier check for each measurement

fig, axes = plt.subplots(2, 2,figsize=(10,10))
fig.suptitle("Outliers by subspecies measurements")

sns.boxplot(
    data=df_raw,
    x="Species",
    y="PetalLengthCm",
    ax=axes[0,0]
)
axes[0,0].set_title("Petal Length by species")

sns.boxplot(
    data=df_raw,
    x="Species",
    y="PetalWidthCm",
    ax=axes[0,1]
)
axes[0,1].set_title("Petal Width by species")

sns.boxplot(
    data=df_raw,
    x="Species",
    y="SepalLengthCm",
    ax=axes[1,0]
)
axes[1,0].set_title("Sepal Length by species")


sns.boxplot(
    data=df_raw,
    x="Species",
    y="SepalWidthCm",
    ax=axes[1,1]
)
axes[1,1].set_title("Sepal Width by species")

plt.tight_layout()
plt.show()




```

* We'll engineer a useful feature, the area of the two plant attributes (sepal and petal), which is a physical feature we can use to attempt identify the subspeices.

```python
# In mwatching sequence
df_raw["SepalArea"] = (df_raw["SepalLengthCm"] * df_raw["SepalWidthCm"])
df_raw["PetalArea"] = (df_raw["PetalLengthCm"] * df_raw["PetalWidthCm"])


df_raw.head(10)
```

There isn't anything else that really needs to be done and theres no other useful features to derive. You could reformat the species to be a '1, 2, 3' or 'a, b, c' but I don't it's feel necessary.

We'll inspect the cleaning before loading this dataframe into seperate .csv.

```python
# Simple copy/paste from above

print("Initial shape:", df_raw.shape)
print('=' * 50)
print("Data types:\n", df_raw.dtypes)
print('=' * 50)
print("missing value count:\n", df_raw.isnull().sum())
print('=' * 50)
print("Duplicates count:\n", df_raw.duplicated().sum())
print('=' * 50)
df_raw.describe()
```

```python
# We're now getting duplicates due to the new columns, not an issue.

print("Duplicates: ",df_raw["SepalArea"].duplicated().sum())
print("Duplicates: ",df_raw["PetalArea"].duplicated().sum())

# You can see the first two rows have a petal area of '0.27999999999999997'
df_raw.head()
```
